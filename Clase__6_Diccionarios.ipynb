##Diccionarios
Son una estructura de datos que almacena valores

sintaxis

```
Nombre_Diccionario={"llave":valor}
```
* La llave **SIEMPRE** tiene que ser alfanumérica
* Valor: cualquier tipo de dato



````python
liga_justicia={"superhéroes":["Batman","Superman","Wonderwoman"],
               "identidades":["Bruce Wayne","Clark Kent","Diana Prince"]}
print(liga_justicia)
````

Los diccionarios **NO** son indexados

````python
liga_justicia["superhéroes"]
````

Estos se identifican mediante llaves

````python
print(liga_justicia["identidades"][2])
````

Las llaves si son indexadas

##Funciones asociadas a los diccionarios

#Clear
Esta funcion se encarga de limpiar toda la informacion de la lista. Esta funcion se escribe de la siguiente forma:

````python
liga_justicia.clear()
print(liga_justicia)
````

#update
Esta funcion perimte modificar el diccionario original agregando nuevos diccionarios al original. Esta funcion se escribe de la siguiente manera:

````python
liga_justicia.update({"Vivos":["si","no","si"]})
print(liga_justicia)
````

##Copy
Esta funcion permite crear un nuevo diccionario con las mismas caracteristicas del original, no obstante las modificaciones de este son indepentientes a las del original. Esta funcion se escribe de la siguiente manera:

````python
Copia = liga_justicia.copy()
Copia.update({"Vivos":["si","no","si"]})
print(liga_justicia)
print(Copia)
````

##Fromkeys
Esta funcion se encarga de asignar un valor preestablecido a un grupo de claves que se encuentran en un diccionario, creando uno nuevo en el proceso, en caso de no asignar un valor el dato de la clave se asignara automaticamente como none.Esta funcion se escribe de la siguiente forma:


````python
materias={"Calculo","Mecanica","Biologia"}
materias_none = dict.fromkeys(materias)
print(materias_none)
Valor = 4
materias_4 = dict.fromkeys(materias,Valor)
print(materias_4)
````

##Get
Esta funcion se encarga de recuperar el valor de una clave, en caso de no existir la clave soliciada esta funcion aarrojara el valor de none, esto permite continuar un codigo pese a presentarse errores. La funcion se escribe de la siguiente manera:

````python
mi_diccionario = {'nombre': 'Ana', 'edad': 30}
valor_edad = mi_diccionario.get('edad') # Devuelve 30
print(valor_edad)
valor_ciudad = mi_diccionario.get('ciudad') # Devuelve None
print(valor_ciudad)
````

##Items
Esta funcion se encarga de permitir la impresion de los items "valor,clave" de un diccionario de manera eficiente con el bucle "for" (Esto se debe a que combierte a estos en tuplas, lo cual puede ser leido por el bucle y permite la imprecion de manera sencilla). La funcion se escribe de la siguiente forma:

````python
mi_diccionario = {
    "nombre": "Ana",
    "edad": 30,
    "ciudad": "Madrid"
}
for clave, valor in mi_diccionario.items():
    print(f"La clave es '{clave}' y el valor es '{valor}'")
````
